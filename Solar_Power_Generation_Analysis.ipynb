```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
generation_1 = pd.read_csv("Plant_1_Generation_Data.csv")
generation_1.head()
```

```python
weather_1 = pd.read_csv("Plant_1_Weather_Sensor_Data.csv")
generation_2 = pd.read_csv("Plant_2_Generation_Data.csv")
weather_2 = pd.read_csv("Plant_2_Weather_Sensor_Data.csv")
print("All four datasets loaded successfully")
```

```python
print("Plant 1 Generation:", generation_1.shape)
print("Plant 1 Weather:", weather_1.shape)
print("Plant 2 Generation:", generation_2.shape)
print("Plant 2 Weather:", weather_2.shape)
```

```python
print("Plant 1 Generation Columns:")
print(generation_1.columns)

print("\nPlant 1 Weather Columns:")
print(weather_1.columns)

print("\nPlant 2 Generation Columns:")
print(generation_2.columns)

print("\nPlant 2 Weather Columns:")
print(weather_2.columns)
```

```python
print("Plant 1 Generation - Missing Values")
print(generation_1.isnull().sum())

print("\nPlant 1 Weather - Missing Values")
print(weather_1.isnull().sum())

print("\nPlant 2 Generation - Missing Values")
print(generation_2.isnull().sum())

print("\nPlant 2 Weather - Missing Values")
print(weather_2.isnull().sum())
```

```python
print("Plant 1 Generation Duplicates:", generation_1.duplicated().sum())

print("Plant 1 Weather Duplicates:", weather_1.duplicated().sum())

print("Plant 2 Generation Duplicates:", generation_2.duplicated().sum())

print("Plant 2 Weather Duplicates:", weather_2.duplicated().sum())
```

```python
print("Plant 1 Generation Data Types")
print(generation_1.dtypes)

print("\nPlant 1 Weather Data Types")
print(weather_1.dtypes)

print("\nPlant 2 Generation Data Types")
print(generation_2.dtypes)

print("\nPlant 2 Weather Data Types")
print(weather_2.dtypes)
```

```python
# Convert DATE_TIME to datetime format

generation_1['DATE_TIME'] = pd.to_datetime(generation_1['DATE_TIME'])
weather_1['DATE_TIME'] = pd.to_datetime(weather_1['DATE_TIME'])
generation_2['DATE_TIME'] = pd.to_datetime(generation_2['DATE_TIME'])
weather_2['DATE_TIME'] = pd.to_datetime(weather_2['DATE_TIME'])
print("Date conversion completed successfully")
```

```python
print("Plant 1 Generation:", generation_1['DATE_TIME'].dtype)
print("Plant 1 Weather:", weather_1['DATE_TIME'].dtype)
print("Plant 2 Generation:", generation_2['DATE_TIME'].dtype)
print("Plant 2 Weather:", weather_2['DATE_TIME'].dtype)
```

```python
print("Plant 1 - Generation Data")
display(generation_1.describe())

print("Plant 1 - Weather Data")
display(weather_1.describe())

print("Plant 2 - Generation Data")
display(generation_2.describe())

print("Plant 2 - Weather Data")
display(weather_2.describe())
```

```python
# Calculate average power for Plant 1
plant1_avg_dc = generation_1['DC_POWER'].mean()
plant1_avg_ac = generation_1['AC_POWER'].mean()

# Calculate average power for Plant 2
plant2_avg_dc = generation_2['DC_POWER'].mean()
plant2_avg_ac = generation_2['AC_POWER'].mean()

print("Plant 1 Average DC Power:", round(plant1_avg_dc, 2))
print("Plant 1 Average AC Power:", round(plant1_avg_ac, 2))
print("Plant 2 Average DC Power:", round(plant2_avg_dc, 2))
print("Plant 2 Average AC Power:", round(plant2_avg_ac, 2))
```

```python
plants = ['Plant 1', 'Plant 2']
avg_dc = [generation_1['DC_POWER'].mean(),generation_2['DC_POWER'].mean()]
avg_ac = [generation_1['AC_POWER'].mean(),generation_2['AC_POWER'].mean()]
x = np.arange(len(plants))
width = 0.35
plt.figure(figsize=(8, 5))
plt.bar(x - width/2, avg_dc, width, label='Average DC Power')
plt.bar(x + width/2, avg_ac, width, label='Average AC Power')
plt.xticks(x, plants)
plt.ylabel('Average Power')
plt.title('Average Power Comparison: Plant 1 vs Plant 2')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Merge Plant 1 generation data with weather data

plant1_data = pd.merge(generation_1,weather_1,on='DATE_TIME',how='inner')
plant1_data.head()
```

```python
# Filter data with positive solar irradiation
daytime_data = plant1_data[plant1_data['IRRADIATION'] > 0]

# Create scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(daytime_data['IRRADIATION'],daytime_data['DC_POWER'],alpha=0.1,s=10)
plt.xlabel('Solar Irradiation')
plt.ylabel('DC Power')
plt.title('Solar Irradiation vs DC Power - Plant 1 (Daytime)')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(plant1_data['MODULE_TEMPERATURE'],plant1_data['DC_POWER'],alpha=0.1,s=10)
plt.xlabel('Module Temperature')
plt.ylabel('DC Power')
plt.title('Module Temperature vs DC Power - Plant 1')
plt.show()
```

```python
# Selected important columns
correlation_data = plant1_data[['IRRADIATION', 'MODULE_TEMPERATURE', 'DC_POWER']]

#correlation
correlation_matrix = correlation_data.corr()

# correlation heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(correlation_matrix,annot=True,cmap='Greens')
plt.title('Correlation Analysis - Plant 1')
plt.show()
```

```python
# Extract date from DATE_TIME
plant1_data['DATE'] = plant1_data['DATE_TIME'].dt.date

# Calculate daily average DC power
daily_power = plant1_data.groupby('DATE')['DC_POWER'].mean()

# Plot daily average power
plt.figure(figsize=(12, 5))
plt.plot(daily_power.index, daily_power.values, marker='o')
plt.xlabel('Date')
plt.ylabel('Average DC Power')
plt.title('Daily Average DC Power Generation - Plant 1')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Extract date from both datasets
generation_1['DATE'] = generation_1['DATE_TIME'].dt.date
generation_2['DATE'] = generation_2['DATE_TIME'].dt.date

# Calculate daily average DC power
daily_power_1 = generation_1.groupby('DATE')['DC_POWER'].mean()
daily_power_2 = generation_2.groupby('DATE')['DC_POWER'].mean()

# Plot both plants
plt.figure(figsize=(12, 5))
plt.plot(daily_power_1.index, daily_power_1.values, label='Plant 1', marker='o')
plt.plot(daily_power_2.index, daily_power_2.values, label='Plant 2', marker='o')
plt.xlabel('Date')
plt.ylabel('Average DC Power')
plt.title('Daily Average DC Power: Plant 1 vs Plant 2')
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Calculate average DC power for both plants
plant1_avg_power = generation_1['DC_POWER'].mean()
plant2_avg_power = generation_2['DC_POWER'].mean()

# Find maximum DC power for both plants
plant1_max_power = generation_1['DC_POWER'].max()
plant2_max_power = generation_2['DC_POWER'].max()

#Results
print("Plant 1 Average DC Power:", round(plant1_avg_power, 2))
print("Plant 2 Average DC Power:", round(plant2_avg_power, 2))

print("Plant 1 Maximum DC Power:", round(plant1_max_power, 2))
print("Plant 2 Maximum DC Power:", round(plant2_max_power, 2))
```
